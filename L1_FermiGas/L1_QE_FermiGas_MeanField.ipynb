## Data
The files in `data/` are the complete archive files of each nucleus: columns `Z A E[GeV] θ[deg] ω[GeV] dσ/dΩdω[nb/sr/GeV] error citation`.
Each (E, θ) pair is one spectrum. Panels A and C use carbon (`12C.dat`); Panel B also uses ⁴He, ¹⁶O, ²⁷Al, ⁴⁰Ca, ⁵⁶Fe, ¹⁹⁷Au and ²⁰⁸Pb.

## A. Scaling: one curve for all kinematics
Divide each spectrum by the Mott cross section and by the single-nucleon factors $G_{L,T}$ (which carry all the form-factor and
kinematic dependence of free-nucleon scattering, and the $1/k_F^3$ normalisation of the Fermi sphere), and plot against the scaling variable
$$\psi' = \frac{\lambda-\tau}{\sqrt{\xi_F\,[(1+\lambda)\tau+\kappa\sqrt{\tau(1+\tau)}]}},\qquad
\lambda=\frac{\omega-E_{\rm shift}}{2M},\ \kappa=\frac{q}{2M},\ \tau=\kappa^2-\lambda^2 .$$
If scattering happens on individual nucleons with a universal momentum distribution, the result $f(\psi')$ is **one function**, independent
of $q$ (Donnelly & Sick). For the Fermi gas it is the parabola $f=\frac34(1-\psi'^2)$. Data from $q\approx300$ to $1000$ MeV
collapse remarkably well on the low-$\omega$ side ($\psi'<0$) — but onto an *asymmetric* curve with a tail, not a parabola. Everything the Fermi gas
is missing (the real momentum distribution, binding, correlations) hides in that shape. Beyond $\psi'\gtrsim1$ scaling breaks: Δ excitation and two-body currents enter.

````python
demo.show_scaling()
````

## B. Superscaling: one curve for different nuclei
Panel A showed **scaling of the first kind**: for one nucleus, spectra at very different kinematics collapse onto one function of $\psi'$.
**Scaling of the second kind** asks whether that function is also the *same for all nuclei*. The only nuclear scale in the scaling variable is the Fermi momentum $k_F$:
$\psi'$ is measured in units of $k_F$, and $f(\psi')$ is multiplied by $k_F$. When both kinds of scaling hold, one speaks of **superscaling** (Donnelly and Sick, 1999).

**The two settings of the panel.**
- *each nucleus: its own Fermi momentum* — every nucleus is scaled with its own $k_F$ and $E_{\rm shift}$ (listed in the legend). These are the values fitted by
Maieron, Donnelly and Sick, Phys. Rev. C 65 (2002) 025502, Table 1; ⁴He and ¹⁶O are not in their table, and the values marked with * are our guesses.
In this setting the two "common" sliders are switched off.
- *all nuclei: one common Fermi momentum* — every selected nucleus is scaled with the **same** $k_F$ and $E_{\rm shift}$, set by the two sliders. This is the test:
if one Fermi momentum were enough for all nuclei, the curves would still lie on top of each other. They do not — light and heavy nuclei separate. Move the slider and watch
which nucleus can be brought onto the curve, and that the others then move away.

**Try this.** (1) Start with ¹²C, ⁴⁰Ca and ²⁰⁸Pb, switch to one common Fermi momentum: which nucleus moves away, and in which direction? (2) Back to own values: does the collapse work
equally well for $\psi'<0$ and $\psi'>0$? Why would you expect the region $\psi'>0$ to break first? (3) Add ⁴He and find a common Fermi momentum that puts it on the carbon curve. Is a Fermi gas a sensible picture for four nucleons?

*Heavy nuclei:* for gold and lead the electron is accelerated and focused by the Coulomb field of the nucleus (the potential at the centre of lead is about 25 MeV), which shifts and distorts the spectra, most visibly at low beam energy. The archive collects the cross sections as published, and the experiments treated this differently: the JLab E89-008 gold data are Coulomb-corrected, the other gold and lead data sets are not marked as corrected in the archive notes. Keep this in mind when comparing gold and lead with the light nuclei. The $|q|$ slider keeps only spectra whose quasielastic peak lies in the chosen range; the title shows how many spectra are plotted.

## Inclusive cross-sections in Fermi Gas model

The inclusive cross section is given in general by
$$\frac{d^2\sigma}{d\Omega\,d\omega}=\sigma_{\rm Mott}\Big[\frac{Q^4}{|\mathbf{q}|^4}R_L+\Big(\frac{Q^2}{2|\mathbf{q}|^2}+\tan^2\frac{\theta}{2}\Big)R_T\Big],$$
with the nuclear responses obtained by adding free, on-shell nucleons one by one. Nucleons occupy a Fermi sphere $|\mathbf{p}|<k_F$:
$$R_{L,T}(\omega,q)=\frac{3N}{4\pi k_F^3}\int d^3p\;\theta(k_F-|\mathbf{p}|)\,\theta(|\mathbf p+\mathbf{q}|-k_F)\,
\frac{M^2}{E_{\mathbf p}E_{\mathbf p+\mathbf q}}\,w_{L,T}(p,q)\,\delta\big(\omega-\varepsilon_B-E_{\mathbf p+\mathbf q}+E_{\mathbf p}\big).$$
The second $\theta$ is **Pauli blocking**; $\varepsilon_B$ is a constant removal energy (Moniz). The single-nucleon tensor is
$w^{\mu\nu}=-W_1\tilde g^{\mu\nu}+W_2\tilde p^\mu\tilde p^\nu/M^2$, $W_1=\tau G_M^2$, $W_2=(G_E^2+\tau G_M^2)/(1+\tau)$.
The $\delta$-function fixes the angle between $\mathbf p$ and $\mathbf q$, so only a one-dimensional integral over $|\mathbf{p}|$ remains.

*The code is in `qe_l1.py`, next to this notebook.*

## C. Fit the Fermi gas
**What sets the peak.** The position follows from two-body kinematics on a nucleon at rest, $Q^2 = 2M\omega$, shifted by the removal energy;
the width is Fermi motion, $\Delta\omega \simeq 2 q k_F/M$. The coloured areas show the longitudinal and transverse parts separately —
the neutrino cross section inherits the same structure, plus axial responses (Panel D).

**The panel.** Two parameters: the Fermi momentum $k_F$ and the removal energy $\varepsilon_B$. Watch what each one does, then switch data set:
do the *same* $k_F$ and $\varepsilon_B$ work at other kinematics? 
The bottom strip shows data/model: the Fermi gas undershoots the high-$\omega$ side (correlations, two-body currents, Δ) and overshoots below the peak at low $q$. Pauli blocking only matters for $q<2k_F$: switch it off on data sets 8 (320 MeV, 36°) or 11 (400 MeV, 36°) and watch the low-$\omega$ side.
FWHM = full width at half maximum of the quasielastic peak

## D. Neutrino scattering in Fermi gas
The same nuclear model. The lepton tensor is now $V\!-\!A$ and the nucleon current acquires the axial form factor,
$\Gamma^\mu = \gamma^\mu F_1 + \frac{i\sigma^{\mu\nu}q_\nu}{2M}F_2 - \gamma^\mu\gamma_5 F_A - \frac{q^\mu}{M}\gamma_5 F_P$,
so the hadronic tensor has five independent pieces instead of two:
$$\frac{d^2\sigma}{d\Omega\,d\omega}=\sigma_0\Big[V_{CC}R_{CC}+2V_{CL}R_{CL}+V_{LL}R_{LL}+V_TR_T\pm2V_{T'}R_{T'}\Big],$$
with $R_{CC}=W^{00}$, $R_{CL}=-\tfrac12(W^{03}+W^{30})$, $R_{LL}=W^{33}$, $R_T=W^{11}+W^{22}$, $R_{T'}=-\tfrac{i}{2}(W^{12}-W^{21})$.
$R_{T'}$ is the vector–axial interference and flips sign between neutrinos and antineutrinos — that is the whole origin of the $\nu/\bar\nu$ asymmetry.
