## Physics engine

Nucleons occupy a Fermi sphere $|\mathbf{p}|<k_F$. The inclusive cross section is
$$\frac{d^2\sigma}{d\Omega\,d\omega}=\sigma_{\rm Mott}\Big[\frac{Q^4}{|\mathbf{q}|^4}R_L+\Big(\frac{Q^2}{2|\mathbf{q}|^2}+\tan^2\frac{\theta}{2}\Big)R_T\Big],$$
with the nuclear responses obtained by adding free, on-shell nucleons one by one:
$$R_{L,T}(\omega,q)=\frac{3N}{4\pi k_F^3}\int d^3p\;\theta(k_F-|\mathbf{p}|)\,\theta(|\mathbf p+\mathbf{q}|-k_F)\,
\frac{M^2}{E_{\mathbf p}E_{\mathbf p+\mathbf q}}\,w_{L,T}(p,q)\,\delta\big(\omega-\varepsilon_B-E_{\mathbf p+\mathbf q}+E_{\mathbf p}\big).$$
The second $\theta$ is **Pauli blocking**; $\varepsilon_B$ is a constant removal energy (Moniz). The single-nucleon tensor is the elastic one,
$w^{\mu\nu}=-W_1\tilde g^{\mu\nu}+W_2\tilde p^\mu\tilde p^\nu/M^2$, $W_1=\tau G_M^2$, $W_2=(G_E^2+\tau G_M^2)/(1+\tau)$.
The $\delta$-function fixes the angle between $\mathbf p$ and $\mathbf q$, so only a one-dimensional integral over $|\mathbf{p}|$ remains —
fast enough to run under a slider. *(Checked against the closed-form RFG responses of Donnelly & Sick to 0.2%.)*

*The code for all of this is in `qe_l1.py`, next to this notebook.*

## C. Scaling: one curve for all kinematics
Divide each spectrum by the Mott cross section and by the single-nucleon factors $G_{L,T}$ (which carry all the form-factor and
kinematic dependence of free-nucleon scattering, and the $1/k_F^3$ normalisation of the Fermi sphere), and plot against the scaling variable
$$\psi' = \frac{\lambda-\tau}{\sqrt{\xi_F\,[(1+\lambda)\tau+\kappa\sqrt{\tau(1+\tau)}]}},\qquad
\lambda=\frac{\omega-E_{\rm shift}}{2M},\ \kappa=\frac{q}{2M},\ \tau=\kappa^2-\lambda^2 .$$
If scattering happens on individual nucleons with a universal momentum distribution, the result $f(\psi')$ is **one function**, independent
of $q$ (Donnelly & Sick). For the RFG it is the parabola $f=\frac34(1-\psi'^2)$. Data from $q\approx300$ to $1000$ MeV
collapse remarkably well on the low-$\omega$ side ($\psi'<0$) — but onto an *asymmetric* curve with a tail, not a parabola. Everything the Fermi gas
is missing (the real momentum distribution, binding, correlations) hides in that shape. Beyond $\psi'\gtrsim1$ scaling breaks: Δ excitation and two-body currents.

```python
demo.show_C()
```

## C′. Superscaling: one curve for different nuclei
Panel C showed **scaling of the first kind**: for one nucleus, spectra at very different kinematics collapse onto one function of $\psi'$.
**Scaling of the second kind** asks whether that function is also the *same for all nuclei*. In the scaling variable the only nuclear scale is the Fermi momentum $k_F$:
$\psi'$ is measured in units of $k_F$, and $f(\psi')$ is multiplied by $k_F$. If each nucleus is given its own $k_F$, the light and the heavy nuclei should lie on top of each other.
When both kinds of scaling hold, one speaks of **superscaling** (Donnelly and Sick, 1999).

The $k_F$ and $E_{\rm shift}$ values of each nucleus are the ones fitted by Maieron, Donnelly and Sick, Phys. Rev. C 65 (2002) 025502 (Table 1).
⁴He and ¹⁶O are not in their table: the values marked with * are our starting guesses.

## D. Binding is not a number: the mean-field potential
In a transport picture (GiBUU) nucleons move in a density- **and momentum-dependent** potential (Welke *et al.*; parameters by Teis;
Leitner's thesis Eq. 7.17, Table 7.3):
$$V_N(\mathbf p,\mathbf r)=A\frac{\rho}{\rho_0}+B\Big(\frac{\rho}{\rho_0}\Big)^{\tau}
+\frac{2C}{\rho_0}\int\!\frac{d^3p'}{(2\pi)^3}\,\frac{g\,[f_n(\mathbf r,\mathbf p')+f_p(\mathbf r,\mathbf p')]}{1+(\mathbf p-\mathbf p')^2/\Lambda^2}.$$
The struck nucleon enters with $V_N(p\lesssim k_F)$ and leaves with $V_N(|\mathbf p+\mathbf{q}|)$. Energy conservation then reads
$\omega = E_{p'}-E_p + \big[V_N(p')-V_N(p)\big]$: the bracket plays the role of $\varepsilon_B$ — and it depends on $q$ and on where in the nucleus the
interaction happened. With EQS 1 (no momentum dependence) the bracket vanishes: a constant well shifts nothing, which is exactly why Fermi-gas
generators need a separate $\varepsilon_B$ by hand. The right-hand panel uses the LDA: the $^{12}$C density profile sets a local $k_F(r)$ and a local $V_N$.

```python
demo.show_D()
```

## E. The same Fermi gas, now hit by a neutrino
Nothing nuclear changes. The lepton tensor is now $V\!-\!A$ and the nucleon current acquires the axial form factors you saw yesterday,
$\Gamma^\mu = \gamma^\mu F_1 + \frac{i\sigma^{\mu\nu}q_\nu}{2M}F_2 - \gamma^\mu\gamma_5 F_A - \frac{q^\mu}{M}\gamma_5 F_P$,
so the hadronic tensor has five independent pieces instead of two:
$$\frac{d^2\sigma}{d\Omega\,d\omega}=\sigma_0\Big[V_{CC}R_{CC}+2V_{CL}R_{CL}+V_{LL}R_{LL}+V_TR_T\pm2V_{T'}R_{T'}\Big],$$
with $R_{CC}=W^{00}$, $R_{CL}=-\tfrac12(W^{03}+W^{30})$, $R_{LL}=W^{33}$, $R_T=W^{11}+W^{22}$, $R_{T'}=-\tfrac{i}{2}(W^{12}-W^{21})$.
The three "longitudinal" ones are tied together by current conservation for the vector part only; $R_{T'}$ is the vector–axial interference and flips sign
between neutrinos and antineutrinos — that is the whole origin of the $\nu/\bar\nu$ asymmetry. The nucleus below is the same RFG as in Panel B.
*(Single-nucleon tensor from the exact Dirac traces, checked against Llewellyn Smith's free-nucleon cross section.)*

```python
demo.show_E()
```

## F. What the potential does to the response
Put the mean field of Panel D into the Fermi gas. The potential enters **only through energy conservation**,
$$\omega = E_{\mathbf p+\mathbf q}-E_{\mathbf p}+\big[V(|\mathbf p+\mathbf q|)-V(|\mathbf p|)\big],$$
the current is untouched (free spinors, effective four-momentum transfer). Four prescriptions: no potential; a constant well felt before *and* after the vertex
(nothing happens — both energies shift equally); the Moniz prescription (the well acts on the initial nucleon only, i.e. the final nucleon is free: a shift by $\varepsilon_B$);
the momentum-dependent GiBUU potential (a shift that grows with $q$, and a broader peak because of the effective mass). Because the potential acts on the kinematics,
**all responses — electron L and T, and the five neutrino responses — move together.**

```python
demo.show_F()
```
